```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# Step 1: Load dataset
data = pd.read_csv('gdp_data.csv')

# Step 2: Rename columns (if needed)
data = data.rename(columns={
    'GDP_USD': 'GDP',
    'Imports_Pct_GDP': 'Imports_pct',
    'Exports_Pct_GDP': 'Exports_pct',
    'Inflation': 'Inflation',
    'USDINR': 'USDINR',
    'Year': 'Year'
})

# Step 3: Feature engineering - convert percentages back to USD
data['Exports'] = (data['Exports_pct'] / 100) * data['GDP']
data['Imports'] = (data['Imports_pct'] / 100) * data['GDP']

# Step 4: Drop missing or invalid data
cols_needed = ['GDP', 'Exports', 'Imports', 'Inflation', 'USDINR']
data[cols_needed] = data[cols_needed].apply(pd.to_numeric, errors='coerce')
data = data.dropna(subset=cols_needed)

# Step 5: Log-transform the GDP
data['Log_GDP'] = np.log(data['GDP'])

# Step 6: Define X and y
X = data[['Exports', 'Imports', 'Inflation', 'USDINR']]
y = data['Log_GDP']

# Step 7: Add constant
X = sm.add_constant(X)

# Step 8: Train/test split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=42)

# Step 9: Fit the model
model = sm.OLS(y_train, X_train).fit()

# Step 10: Print summary
print(model.summary())

# Step 11: Predict
y_pred_log = model.predict(X_test)

# Convert predictions back from log scale
y_pred = np.exp(y_pred_log)
y_test_actual = np.exp(y_test)

# Step 12: Evaluate
mse = mean_squared_error(y_test_actual, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_actual, y_pred)
print("ols regression result")
print(f"\nTest MSE: {mse:.2f}")
print(f"Test RMSE: {rmse:.2f}")
print(f"Test R² (R-squared): {r2:.4f}")

# Step 13: Plot Actual vs Predicted GDP
plt.figure(figsize=(8, 6))
plt.scatter(y_test_actual, y_pred, color='navy', alpha=0.7, label='Predicted vs Actual')
plt.plot([y_test_actual.min(), y_test_actual.max()], [y_test_actual.min(), y_test_actual.max()],
         color='red', linestyle='--', label='Perfect Fit')
plt.xlabel('Actual GDP (USD)')
plt.ylabel('Predicted GDP (USD)')
plt.title('Actual vs Predicted GDP')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
